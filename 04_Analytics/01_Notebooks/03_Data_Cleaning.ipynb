# Data Cleaning 

### Objective
Prepare the raw e-commerce dataset for analysis by cleaning, validating, and transforming the data.

```python
import numpy as np
import pandas as pd
```

```python
customers = pd.read_csv("../../02_Data/01_Raw_Data/olist_customers_dataset.csv")
orders = pd.read_csv("../../02_Data/01_Raw_Data/olist_orders_dataset.csv")
order_items = pd.read_csv("../../02_Data/01_Raw_Data/olist_order_items_dataset.csv")
products = pd.read_csv("../../02_Data/01_Raw_Data/olist_products_dataset.csv")
payments = pd.read_csv("../../02_Data/01_Raw_Data/olist_order_payments_dataset.csv")
reviews = pd.read_csv("../../02_Data/01_Raw_Data/olist_order_reviews_dataset.csv")
sellers = pd.read_csv("../../02_Data/01_Raw_Data/olist_sellers_dataset.csv")
```

## 1. Orders Dataset

### 1.1 Data Quality Check

```python
print("=" * 60)
print("ORDERS DATA QUALITY")
print("=" * 60)

# Shape
print(f"\nShape: {orders.shape}")

# Data Types
print("\nData Types:")
print(orders.dtypes)

# Missing Values
print("\nMissing Values:")
print(orders.isnull().sum())

# Duplicate Rows
print(f"\nDuplicate Rows: {orders.duplicated().sum()}")

# Order Status
print("\nOrder Status Distribution:")
print(orders["order_status"].value_counts())

# Business Validation
print("\nMissing Values by Order Status:")
print(
    orders.groupby("order_status")[
        [
            "order_approved_at",
            "order_delivered_carrier_date",
            "order_delivered_customer_date",
        ]
    ].apply(lambda x: x.isnull().sum())
)
```

### 1.2 Cleaning

```python
# Remove duplicate rows (if any)
orders = orders.drop_duplicates()

# Business Decision:
# Keep missing delivery-related dates because they belong to
# cancelled, unavailable, or in-progress orders.
# No imputation is performed.
```

### 1.3 Validation

```python
print("=" * 60)
print("ORDERS VALIDATION")
print("=" * 60)

print(f"\nFinal Shape: {orders.shape}")

print("\nMissing Values:")
print(orders.isnull().sum())

print(f"\nDuplicate Rows: {orders.duplicated().sum()}")
```

### 1.4 Business Insight

- The Orders dataset contains 99,441 records with no duplicate rows.
- Missing values are present only in delivery-related columns.
- These missing values mainly belong to cancelled, unavailable, or in-progress orders and represent valid business scenarios.
- No records were removed during the cleaning process.
- The dataset is clean and ready for feature engineering and analysis.

## 2. Remaining Datasets - Data Quality Check

```python
datasets = {
    "Customers": customers,
    "Order Items": order_items,
    "Products": products,
    "Payments": payments,
    "Reviews": reviews,
    "Sellers": sellers
}

for name, df in datasets.items():
    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print(f"Shape: {df.shape}")
    print(f"Duplicate Rows: {df.duplicated().sum()}")
    print(f"Total Missing Values: {df.isnull().sum().sum()}")

    if df.isnull().sum().sum() > 0:
        print("\nMissing Values by Column:")
        print(df.isnull().sum()[df.isnull().sum() > 0])

    print("\nData Types:")
    print(df.dtypes)

    print("\n")
```

## 3. Data Cleaning

```python
# Remove duplicates
customers = customers.drop_duplicates()
orders = orders.drop_duplicates()
order_items = order_items.drop_duplicates()
products = products.drop_duplicates()
payments = payments.drop_duplicates()
reviews = reviews.drop_duplicates()
sellers = sellers.drop_duplicates()

# Convert date columns to datetime
orders["order_purchase_timestamp"] = pd.to_datetime(orders["order_purchase_timestamp"])
orders["order_approved_at"] = pd.to_datetime(orders["order_approved_at"])
orders["order_delivered_carrier_date"] = pd.to_datetime(orders["order_delivered_carrier_date"])
orders["order_delivered_customer_date"] = pd.to_datetime(orders["order_delivered_customer_date"])
orders["order_estimated_delivery_date"] = pd.to_datetime(orders["order_estimated_delivery_date"])

reviews["review_creation_date"] = pd.to_datetime(reviews["review_creation_date"])
reviews["review_answer_timestamp"] = pd.to_datetime(reviews["review_answer_timestamp"])

print("✅ Data Cleaning Completed")
```

## 4. Validation

```python
datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Products": products,
    "Payments": payments,
    "Reviews": reviews,
    "Sellers": sellers
}

print("=" * 80)
print("DATA VALIDATION REPORT")
print("=" * 80)

for name, df in datasets.items():
    print(f"\n{name}")
    print("-" * 40)
    print(f"Shape: {df.shape}")
    print(f"Duplicate Rows: {df.duplicated().sum()}")
    print(f"Total Missing Values: {df.isnull().sum().sum()}")
```

## 5. Key Business Insights

- Orders contain valid missing delivery dates due to cancelled/unavailable orders.
- Customers dataset is complete with no missing values.
- Product and review datasets contain business-driven missing values.
- Duplicate records were removed where necessary.
- The datasets are validated and ready for feature engineering and exploratory data analysis.

```python
import os

# Create cleaned data folder
output_path = '../../02_Data/02_Cleaned_Data'
os.makedirs(output_path, exist_ok=True)

# Save all cleaned datasets
customers.to_csv(f'{output_path}/customers_clean.csv', index=False)
orders.to_csv(f'{output_path}/orders_clean.csv', index=False)
order_items.to_csv(f'{output_path}/order_items_clean.csv', index=False)
products.to_csv(f'{output_path}/products_clean.csv', index=False)
payments.to_csv(f'{output_path}/payments_clean.csv', index=False)
reviews.to_csv(f'{output_path}/reviews_clean.csv', index=False)
sellers.to_csv(f'{output_path}/sellers_clean.csv', index=False)

print("✅ All cleaned datasets saved!")
```
