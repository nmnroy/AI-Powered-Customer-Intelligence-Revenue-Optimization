### Objective
Prepare raw e-commerce tables for analytical modeling by enforcing datetimes, validating missing values, and generating clean CSV exports.


```
import os
import pandas as pd
import numpy as np

raw_dir = '../../02_Data/01_Raw_Data'
clean_dir = '../../02_Data/02_Cleaned_Data'
os.makedirs(clean_dir, exist_ok=True)

customers = pd.read_csv(os.path.join(raw_dir, 'olist_customers_dataset.csv')).drop_duplicates()
orders = pd.read_csv(os.path.join(raw_dir, 'olist_orders_dataset.csv')).drop_duplicates()
order_items = pd.read_csv(os.path.join(raw_dir, 'olist_order_items_dataset.csv')).drop_duplicates()
products = pd.read_csv(os.path.join(raw_dir, 'olist_products_dataset.csv')).drop_duplicates()
payments = pd.read_csv(os.path.join(raw_dir, 'olist_order_payments_dataset.csv')).drop_duplicates()
reviews = pd.read_csv(os.path.join(raw_dir, 'olist_order_reviews_dataset.csv')).drop_duplicates()
sellers = pd.read_csv(os.path.join(raw_dir, 'olist_sellers_dataset.csv')).drop_duplicates()

# Datetime enforcement
order_date_cols = ['order_purchase_timestamp', 'order_approved_at', 
                   'order_delivered_carrier_date', 'order_delivered_customer_date', 
                   'order_estimated_delivery_date']

for col in order_date_cols:
    orders[col] = pd.to_datetime(orders[col])

reviews['review_creation_date'] = pd.to_datetime(reviews['review_creation_date'])
reviews['review_answer_timestamp'] = pd.to_datetime(reviews['review_answer_timestamp'])

print("✅ Timestamps converted to datetime objects successfully.")
```

```
# Export clean datasets
customers.to_csv(os.path.join(clean_dir, 'customers_clean.csv'), index=False)
orders.to_csv(os.path.join(clean_dir, 'orders_clean.csv'), index=False)
order_items.to_csv(os.path.join(clean_dir, 'order_items_clean.csv'), index=False)
products.to_csv(os.path.join(clean_dir, 'products_clean.csv'), index=False)
payments.to_csv(os.path.join(clean_dir, 'payments_clean.csv'), index=False)
reviews.to_csv(os.path.join(clean_dir, 'reviews_clean.csv'), index=False)
sellers.to_csv(os.path.join(clean_dir, 'sellers_clean.csv'), index=False)

print(f"✅ All 7 cleaned datasets successfully written to {clean_dir}")
```
