### Objective
Inspect raw tables, verify schemas, identify missing values and duplicate rows, and generate an initial data quality summary.


```
import os
import pandas as pd
import numpy as np

raw_dir = '../../02_Data/01_Raw_Data'

customers = pd.read_csv(os.path.join(raw_dir, 'olist_customers_dataset.csv'))
orders = pd.read_csv(os.path.join(raw_dir, 'olist_orders_dataset.csv'))
order_items = pd.read_csv(os.path.join(raw_dir, 'olist_order_items_dataset.csv'))
products = pd.read_csv(os.path.join(raw_dir, 'olist_products_dataset.csv'))
payments = pd.read_csv(os.path.join(raw_dir, 'olist_order_payments_dataset.csv'))
reviews = pd.read_csv(os.path.join(raw_dir, 'olist_order_reviews_dataset.csv'))
sellers = pd.read_csv(os.path.join(raw_dir, 'olist_sellers_dataset.csv'))

datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Products": products,
    "Payments": payments,
    "Reviews": reviews,
    "Sellers": sellers
}

for name, df in datasets.items():
    print(f"{name:15}: {df.shape[0]:>7,} rows | {df.shape[1]:>2} columns")
```

### Data Quality & Profiling Summary
Audit each table for missing values and duplicates to establish a baseline before data cleaning.


```
summary = []

for name, df in datasets.items():
    summary.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

summary_df = pd.DataFrame(summary)
os.makedirs('../../04_Analytics/03_Outputs', exist_ok=True)
summary_df.to_csv('../../04_Analytics/03_Outputs/01_Data_Profiling_Summary.csv', index=False)
summary_df
```

### Key Observations:
- **Zero duplicate rows** detected across all primary tables.
- Missing values in `Orders` belong strictly to delivery-related milestones for cancelled/in-transit orders.
- `Customers` contains 99,441 rows with two identifiers: `customer_id` and `customer_unique_id`.
